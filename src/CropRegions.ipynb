```python
import cv2
import os
import glob

# 1. Setup your folders
input_dir = "../data/bill_processing/normalized_images/"
output_dir = "../data/bill_processing/reference_components/"
os.makedirs(output_dir, exist_ok=True)

# Grab all the normalized images you just made
files = glob.glob(os.path.join(input_dir, "norm_*.*"))

print("Starting Component Extraction...")
print("INSTRUCTIONS FOR THE POP-UP WINDOW:")
print("1. Click and drag to draw a box around a distinct feature (e.g., a face or number).")
print("2. Press SPACE or ENTER to confirm that specific box (it will turn green).")
print("3. Draw as many boxes as you need on the current bill.")
print("4. Press ESC when you are finished with the current bill to move to the next.")
print("   (If you make a mistake, press 'c' to clear the current boxes).")
print("-" * 50)

for filepath in files:
    filename = os.path.basename(filepath)
    img = cv2.imread(filepath)
    
    if img is None:
        continue
        
    print(f"Selecting components for: {filename}")
    
    # 2. Open the multi-selection interactive window
    # Remember: This opens outside your browser!
    rois = cv2.selectROIs("Select Components (ENTER to confirm box, ESC to finish image)", img, showCrosshair=True, fromCenter=False)
    
    # 3. Loop through every box you drew on this image
    # 'rois' is an array of [x, y, w, h] coordinates
    for i, rect in enumerate(rois):
        x, y, w, h = rect
        if w > 0 and h > 0:
            # Crop the component
            component = img[y:y+h, x:x+w]
            
            # Create a unique filename for each component
            # Example: norm_clean_500PesosFront_comp_0.png
            base_name = os.path.splitext(filename)[0]
            save_path = os.path.join(output_dir, f"{base_name}_comp_{i}.png")
            
            # Save it
            cv2.imwrite(save_path, component)
            print(f"  -> Saved: {save_path}")

# 4. Clean up
cv2.destroyAllWindows()
print("\nAll reference components extracted and saved!")
```
