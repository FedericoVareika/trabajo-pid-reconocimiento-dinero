```python
import cv2
import os
import glob

input_dir = "../data/bill_processing/normalized_images/"
output_dir = "../data/bill_processing/reference_components/"
os.makedirs(output_dir, exist_ok=True)

files = glob.glob(os.path.join(input_dir, "norm_*.*"))

print("Extracción de componentes de referencia")
print("-" * 40)
print("Instrucciones de la ventana:")
print("- Arrastre el ratón para dibujar un recuadro sobre la característica.")
print("- ENTER o ESPACIO para confirmar el recuadro (se pondrá verde).")
print("- ESC para finalizar la imagen actual y pasar a la siguiente.")
print("- Tecla 'c' para limpiar las selecciones actuales.")
print("-" * 40)

for filepath in files:
    filename = os.path.basename(filepath)
    img = cv2.imread(filepath)
    
    if img is None:
        print(f"Error: No se pudo cargar {filename}")
        continue
        
    print(f"Procesando: {filename}")
    
    # Selección interactiva de múltiples regiones de interés (ROI)
    rois = cv2.selectROIs("Extractor de Componentes", img, showCrosshair=True, fromCenter=False)
    
    if len(rois) > 0:
        base_name = os.path.splitext(filename)[0]
        
        for i, rect in enumerate(rois):
            x, y, w, h = rect
            if w > 0 and h > 0:
                # Recorte y guardado del componente
                component = img[y:y+h, x:x+w]
                save_path = os.path.join(output_dir, f"{base_name}_comp_{i}.png")
                
                cv2.imwrite(save_path, component)
                print(f"  -> Guardado: {save_path}")
    else:
        print(f"  -> No se seleccionaron componentes para {filename}.")

cv2.destroyAllWindows()
print("\nProceso finalizado. Componentes guardados en:", output_dir)
```
