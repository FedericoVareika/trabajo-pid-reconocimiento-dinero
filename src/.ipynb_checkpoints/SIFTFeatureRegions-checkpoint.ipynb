```python
import cv2
import os
import glob
import numpy as np

# 1. Setup your folders
input_dir = "../data/bill_processing/reference_components/"
output_db_dir = "../data/database/sift_database/"      
output_vis_dir = "../data/databasesift_visuals/"      

os.makedirs(output_db_dir, exist_ok=True)
os.makedirs(output_vis_dir, exist_ok=True)

# 2. Initialize the SIFT detector
# SIFT is now built directly into the standard cv2 library!
sift = cv2.SIFT_create()

# Grab all your cropped components
files = glob.glob(os.path.join(input_dir, "*.png"))

print("Extracting SIFT features from components...\n")

for filepath in files:
    filename = os.path.basename(filepath)
    base_name = os.path.splitext(filename)[0]
    
    img = cv2.imread(filepath)
    if img is None:
        continue
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    
    # 3. Detect keypoints and compute descriptors
    # SIFT generates a 128-dimensional vector for each point (SURF did 64)
    kp, des = sift.detectAndCompute(gray, None)
    
    if des is not None:
        # 4. Save the descriptors as a NumPy file (This is your database!)
        db_save_path = os.path.join(output_db_dir, f"{base_name}.npy")
        np.save(db_save_path, des)
        
        # 5. Draw the keypoints for visual confirmation
        img_with_kp = cv2.drawKeypoints(img, kp, None, color=(0, 255, 0), flags=cv2.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS)
        
        vis_save_path = os.path.join(output_vis_dir, f"{base_name}_sift_keypoints.jpg")
        cv2.imwrite(vis_save_path, img_with_kp)
        
        print(f"Extracted {len(kp)} SIFT features from {filename}")
    else:
        print(f"WARNING: No features found in {filename}.")

print("\nSIFT extraction complete! Check the 'sift_visuals' folder.")
```
