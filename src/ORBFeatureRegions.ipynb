```python
import cv2
import os
import glob
import numpy as np

input_dir = "../data/bill_processing/reference_components/"
output_db_dir = "../data/database/orb_database/"
output_vis_dir = "../data/database/orb_visuals/"

os.makedirs(output_db_dir, exist_ok=True)
os.makedirs(output_vis_dir, exist_ok=True)

orb = cv2.ORB_create(nfeatures=2000)

files = glob.glob(os.path.join(input_dir, "*.png"))

print(f"Iniciando extracción ORB desde: {input_dir}\n")

for filepath in files:
    filename = os.path.basename(filepath)
    base_name = os.path.splitext(filename)[0]
    
    img = cv2.imread(filepath)
    if img is None:
        print(f"Error: No se pudo cargar {filename}.")
        continue
        
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    
    # Detección y descripción
    kp, des = orb.detectAndCompute(gray, None)
    
    if des is not None:
        # Almacenamiento de descriptores
        db_save_path = os.path.join(output_db_dir, f"{base_name}.npy")
        np.save(db_save_path, des)
        
        # Generación de visualización técnica
        img_with_kp = cv2.drawKeypoints(
            img, kp, None, 
            color=(0, 255, 0), 
            flags=cv2.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS
        )
        
        vis_save_path = os.path.join(output_vis_dir, f"{base_name}_orb_keypoints.jpg")
        cv2.imwrite(vis_save_path, img_with_kp)
        
        print(f"Procesado: {filename} ({len(kp)} descriptores)")
    else:
        print(f"Aviso: No se detectaron características en {filename}.")

print("\nExtracción finalizada. Datos guardados en la base de datos ORB.")
```
